# 02 · Tablas de datos con pandas

**Objetivos**
* Leer una tabla (CSV) desde internet.
* Filtrar, ordenar, agrupar y graficar datos.

`pandas` es la biblioteca de Python para trabajar con tablas, parecida a una hoja de cálculo pero programable.

````python
import pandas as pd
import matplotlib.pyplot as plt

URL = "https://raw.githubusercontent.com/pesco13/Python-para-Quimica/main/datos/compuestos_industriales.csv"
df = pd.read_csv(URL)
df.head()
````

Cada fila es un compuesto y cada columna un dato. `head()` muestra las primeras 5 filas.

````python
print("Filas y columnas:", df.shape)
print(df.columns.tolist())
df["categoria"].value_counts()
````

## Filtrar

Seleccionar solo los conservantes:

````python
conservantes = df[df["categoria"] == "Conservante"]
conservantes
````

## Agregar una columna calculada

La longitud del SMILES (número de caracteres) es una medida muy burda del tamaño de la molécula; en el notebook 03 calcularemos descriptores reales.

````python
df["largo_smiles"] = df["smiles"].str.len()
df.sort_values("largo_smiles", ascending=False).head()
````

## Agrupar y graficar

````python
conteo = df["categoria"].value_counts()
conteo.plot(kind="barh")
plt.xlabel("Número de compuestos")
plt.title("Compuestos por categoría")
plt.show()
````

## Guardar resultados

`to_csv` guarda la tabla; en Colab aparece en el panel de archivos (ícono de carpeta) y se puede descargar.

````python
df.to_csv("mis_compuestos.csv", index=False)
````
